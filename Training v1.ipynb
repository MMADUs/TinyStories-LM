# Training V1 Notebook

## Import Modules

```python
%load_ext autoreload
%autoreload 2

import pickle

from config import model_config
from train import train_model, eval_model
from dataset import get_dataloaders
from utils import model_summary
from plot import plot_loss, plot_perplexity
```

## Get Dataloaders

```python
train_dl, val_dl, test_dl, tokenizer = get_dataloaders(model_config)
```

## Model Preview

```python
model_summary(model_config, tokenizer, depth=3)
```

## Train Model

```python
history = train_model(
    model_config,
    train_dl,
    val_dl,
    tokenizer,
    version="v1",
    initial_train=True,
)
```

```python
plot_loss(history)
```

```python
plot_perplexity(history)
```

## Evaluate Model

```python
result = eval_model(model_config, "v1", test_dl, tokenizer)
```

```python
print(result)
```
