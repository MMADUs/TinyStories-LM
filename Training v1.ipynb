# Training V1 Notebook

## Import Modules

```python
%load_ext autoreload
%autoreload 2

import pickle

from config import model_config
from train import train_model, eval_model
from dataset import get_dataloaders
from utils import model_summary
from plot import line_plot
```

## Get Dataloaders

```python
train_dl, val_dl, test_dl, tokenizer = get_dataloaders(model_config)
```

## Model Preview

```python
model_summary(model_config, tokenizer, depth=3)
```

## Train Model

```python
history = train_model(
    train_dl,
    val_dl,
    tokenizer,
    model_config,
    version="v1",
    initial_train=True,
)
```

```python
line_plot(
    history["train_loss"],
    history["val_loss"],
    x1_label="Train Loss",
    x2_label="Val Loss",
)
```

```python
line_plot(
    history["train_perplexity"],
    history["val_perplexity"],
    x1_label="Train Perplexity",
    x2_label="Val Perplexity",
)
```

## Evaluate Model

```python
result = eval_model(model_config, "v1", test_dl, tokenizer)
```

```python
line_plot(
    result["test_loss"],
    history["test_perplexity"],
    x1_label="Test Loss",
    x2_label="Test Perplexity",
)
```

```python
# save
with open(".output/history.pkl", "wb") as f:
    pickle.dump(history, f)
```
