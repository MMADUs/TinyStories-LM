# Story Prompts (SFT Inference)

```python
config = load_config(model_40M_params)
loader = ModelLoader(config, stage="finetuning", version="v1_40M", load_from_epoch=1)
```

```python
loader.autoregressive_attribute(
    max_new_tokens=2000,
    temperature=1.0,
    top_k=50,
    top_p=0.95,
)
```

```python
features = ["Dialogue", "BadEnding", "MoralValue"]
words = ["ride", "work", "upset"]
summary = "Lily steals a new bike from a store and gets into an accident while riding it, resulting in her getting hurt and being sent to jail, losing her parents' trust and love. The moral of the story is not to steal."

features, words, summary, story = generate_story(loader, features, words, summary)

print("Features:", features)
print("\n")
print("Words:", words)
print("\n")
print("Prompt:", summary)
print("\n")
print("Story:", story)
```
